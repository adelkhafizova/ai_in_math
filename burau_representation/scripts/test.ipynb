```python
def generate_free_words(length):
    letters = ['A', 'B', 'a', 'b']
    inverse_map = {0: 2, 1: 3, 2: 0, 3: 1}
    current_word = [-1] * length
    pos = 0
    count = 0
    
    while pos >= 0:
        current_word[pos] += 1
        
        if current_word[pos] >= len(letters):
            current_word[pos] = -1
            pos -= 1
            continue
        
        if pos > 0 and current_word[pos] == inverse_map[current_word[pos-1]]:
            continue
        
        if pos == length - 1:
            word = ''.join(letters[i] for i in current_word)
            print(word)
            count += 1
        else:
            pos += 1
    
    print(f"Total words: {count}")

# Run the function
generate_free_words(15)
```
